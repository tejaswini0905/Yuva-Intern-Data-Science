```python
Q1 = df[["Quantity", "Price"]].quantile(0.25)
Q3 = df[["Quantity", "Price"]].quantile(0.75)

IQR = Q3 - Q1

lower = Q1 - 1.5 * IQR
upper = Q3 + 1.5 * IQR

print("Lower bounds:")
print(lower)

print("\nUpper bounds:")
print(upper)
```

```python
quantity_outliers = (df["Quantity"] > upper["Quantity"]).sum()
price_outliers = (df["Price"] > upper["Price"]).sum()

print("Quantity outliers:", quantity_outliers)
print("Price outliers:", price_outliers)
```

```python
print("Highest quantities:")
print(df.nlargest(10, "Quantity")[["Invoice", "StockCode", "Description", "Quantity", "Price"]])

print("\nHighest prices:")
print(df.nlargest(10, "Price")[["Invoice", "StockCode", "Description", "Quantity", "Price"]])
```
